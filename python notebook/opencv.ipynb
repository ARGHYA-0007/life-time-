```python
import cv2
import numpy as np

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        print("there are some problem detected to camera")
        break
    
    # Get frame dimensions
    h, w = frame.shape[:2]
    
    # Calculate rotation matrix for 30 degrees
    center = (w // 2, h // 2)
    rotation_matrix = cv2.getRotationMatrix2D(center, 30, 1.0)
    
    # Apply rotation
    rotated_frame = cv2.warpAffine(frame, rotation_matrix, (w, h))
    
    cv2.imshow("video showing", rotated_frame)
    
    if cv2.waitKey(1) & 0xFF == ord(" "):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break
    
    flipped_frame = cv2.flip(frame, 1)
    gray_image = cv2.cvtColor(flipped_frame, cv2.COLOR_BGR2GRAY)
    _, thres = cv2.threshold(gray_image, 125, 255, cv2.THRESH_BINARY)
    contours, hierarchy = cv2.findContours(thres, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)
    
    for contour in contours:
        area = cv2.contourArea(contour)
        
        # Only process contours with area greater than 500 (adjust as needed)
        if area > 500:
            approx = cv2.approxPolyDP(contour, 0.01 * cv2.arcLength(contour, True), True)
            corners = len(approx)
            
            if corners == 3:
                shape_name = 'triangle'
            elif corners == 4:
                shape_name = 'rectangle'
            elif corners == 5:
                shape_name = 'pentagon'
            elif corners == 6:
                shape_name = 'hexagon'
            else:
                shape_name = "circle"
            
            cv2.drawContours(flipped_frame, [approx], 0, (0, 255, 0), 3)
            x = approx.ravel()[0]
            y = approx.ravel()[1] - 10
            cv2.putText(flipped_frame, shape_name, (x, y), cv2.FONT_HERSHEY_COMPLEX, 0.6, (255, 0, 0), 2)
    
    cv2.imshow("video_contours", flipped_frame)
    if cv2.waitKey(1) & 0xFF == ord(" "):
        break

cap.release()
cv2.destroyAllWindows()
```

## smile,eyes and face detection 

### scaleFactor (e.g., 1.1, 1.3, 1.8): How much the image size is reduced at each scale

### Lower values (1.05-1.2): More thorough search, detects more objects (including false positives)
### Higher values (1.3-2.0): Faster but might miss some objects


### minNeighbors (e.g., 3, 5, 20): How many neighbors each candidate rectangle should have to retain it

### Lower values (3-5): More detections but MORE false positives
### Higher values (15-30): Fewer false positives but might miss real objects

```python
import cv2
frontal_face=cv2.CascadeClassifier(r"C:\Users\arghy\Downloads\haarcascade_frontalface_alt.xml")
smiles=cv2.CascadeClassifier(r"C:\Users\arghy\Downloads\haarcascade_smile.xml")
eyes=cv2.CascadeClassifier(r"C:\Users\arghy\Downloads\haarcascade_eye.xml")
cap=cv2.VideoCapture(0)
while True:
    ret,frame=cap.read()
    flipped_frame=cv2.flip(frame,1)
    gray=cv2.cvtColor(flipped_frame,cv2.COLOR_BGR2GRAY)
    faces=frontal_face.detectMultiScale(gray,1.3,5)
   
    
    for(x,y,w,h) in faces:
        roi_smile=gray[y:y+h,x:x+w]
        roi_color = flipped_frame[y:y+h, x:x+w]
        cv2.rectangle(flipped_frame,(x,y),(x+w,y+h),(255,0,0),2)
        cv2.putText(flipped_frame,"FACE",(x,y-60),cv2.FONT_HERSHEY_COMPLEX,0.6,(255,0,0),3)
        smile=smiles.detectMultiScale(roi_smile,1.8,20)
        eye=eyes.detectMultiScale(roi_smile,1.1,5)
        if len(smile)>0:
            for (sx,sy,sw,sh) in smile: 
                cv2.rectangle(flipped_frame,(x+sx,y+sy),(x+sx+sw,y+sy+sh),(255,255,255),2)
            cv2.putText(flipped_frame,"mouth detected",(x,y-40),cv2.FONT_HERSHEY_COMPLEX,0.6,(0,0,255),2)
        if len(eye)>0:
            for (ex,ey,ew,eh) in eye: 
                cv2.rectangle(flipped_frame,(x+ex,y+ey),(x+ex+ew,y+ey+eh),(255,225,0),2)
            cv2.putText(flipped_frame,"eyes detected",(x,y-20),cv2.FONT_HERSHEY_COMPLEX,0.6,(0,255,0),1)
    if not ret:
        break
    cv2.imshow("video",flipped_frame)
    if cv2.waitKey(1) & 0xFF==ord(" "):
        break
cap.release()
cv2.destroyAllWindows()
```

## we can use xml file without downloading that xml file by under written method


```python
import cv2

# Use OpenCV's data directory
cascade_path = cv2.data.haarcascades + 'haarcascade_frontalface_default.xml'
face = cv2.CascadeClassifier(cascade_path)

# Verify it loaded correctly
if face.empty():
    print("Error: Could not load cascade classifier!")
    exit()

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    
    if not ret:
        break
    
    flipped_frame = cv2.flip(frame, 1)
    gray = cv2.cvtColor(flipped_frame, cv2.COLOR_BGR2GRAY)
    faces = face.detectMultiScale(gray, 1.3, 5)
    
    for (x, y, w, h) in faces:
        cv2.rectangle(flipped_frame, (x, y), (x+w, y+h), (255, 0, 0), 2)
        cv2.putText(flipped_frame, "FACE", (x, y-30), cv2.FONT_HERSHEY_COMPLEX, 0.6, (255, 0, 0), 2)
    
    cv2.imshow("video", flipped_frame)
    
    if cv2.waitKey(1) & 0xFF == ord(" "):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
# cascade_path=cv2.data.haarcascades +"haarcascade_upperbody.xml"
body=cv2.CascadeClassifier(r"C:\Users\arghy\Downloads\haarcascade_upperbody.xml")
cap=cv2.VideoCapture(0)
while True:
    ret,frame=cap.read()
    if not ret:
        break
    flipped_frame=cv2.flip(frame,1)
    gray=cv2.cvtColor(flipped_frame,cv2.COLOR_BGR2GRAY)
    gray = cv2.equalizeHist(gray) # improve contrast
    bodies=body.detectMultiScale(gray,1.1,4)
    for (x,y,w,h) in bodies:
        cv2.rectangle(flipped_frame,(x,y),(x+w,y+h),(0,255,0),2)
        cv2.putText(flipped_frame,"human body detected",(x,y-10),cv2.FONT_HERSHEY_COMPLEX,0.6,(255,0,0),1)
    cv2.imshow("human body detection system",flipped_frame)
    if cv2.waitKey(1) & 0xFF==ord(" "):
        break
cap.release()
cv2.destroyAllWindows()
```
