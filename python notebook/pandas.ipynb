### filling missing data with any thing we want

```python
import pandas as pd

data = {
    'EmployeeID': [101, 102, 103, 104, 105],
    'Name': ['Alice',None, 'Charlie', 'Diana', 'Ethan'],
    'Department': ['HR', 'IT', 'Finance', 'IT', 'HR'],
    'Salary': [50000, 65000, None, 62000, 52000],
    'JoiningDate': pd.to_datetime(['2018-01-10', None, '2017-07-23', '2020-11-01', '2018-05-19']),
    'ExperienceYears': [6, 4, 7, 3, 5],
    'City': [None ,'Los Angeles', 'Chicago', 'Houston', 'Phoenix']
}
df=pd.DataFrame(data)

df.fillna(0,inplace=True) #where is missing data these will be filled by 0
print(df)

#if we wanted to fill missing data with any calculated number like mean
df["Salary"].fillna(df["Salary"].mean(),inplace=True)
print(df)
```

### filling missing data by interpolation 
#### 👉but this dont work for catagorical data like name
#### 👉we use this to avoid dropping row for a missing data
#### 👉this use a predicted data using a same algo so this always will not be true

```python
import pandas as pd

data = {
    'EmployeeID': [101, 102, 103, 104, 105],
    'Name': ['Alice','chaman', 'Charlie', 'Diana', 'Ethan'],
    'Department': ['HR', 'IT', 'Finance', 'IT', 'HR'],
    'Salary': [50000, 65000, None, 62000, 52000],
    'JoiningDate': pd.to_datetime(['2018-01-10', None, '2017-07-23', '2020-11-01', '2018-05-19']),
    'ExperienceYears': [6, None, 7, 3, 5],
    'City': ['kolkata' ,'Los Angeles', 'Chicago', 'Houston', 'Phoenix']
}
df=pd.DataFrame(data)


df.interpolate(method='linear',axis = 0,inplace=True) #there are so many methods like
                                                      #linear,polynomial,time
                    # axis = 0 means interpolation in rows and this is also default
print(df)
```

### merging two different dataframe 
#### innerjoin
#### outerjoin
#### leftjoin
#### rightjoin

```python
import pandas as pd
name={
    'names':['arghya','ani','amit'],
    'id':[1,2,3]
}
df_names=pd.DataFrame(name)
marks={
    'marks':[40,50,60],
    'id':[1,2,4]
}
df_marks=pd.DataFrame(marks)

#innerjoin
print('this is innerjoin')
pd.merge(df_names,df_marks,on='id',how='inner' ) #basically this is the intersection of two data set by a common column name
#outerjoin
print('this is outerjoin')
pd.merge(df_names,df_marks,on='id',how='outer' ) #basically this is the union of two data set and filling missing or unmatched data with NaN
#leftjoin
print('this is leftjoin')
pd.merge(df_names,df_marks,on='id',how='left' ) #left datas are 1st priority then acc to that matching with right data set
#rightjoin
print('this is rightjoin')
pd.merge(df_names,df_marks,on='id',how='right' ) #right datas are 1st priority then acc to that matching with left side data 
```

### concatenation -------> adding rows of columns of two or more data frame

### finding value in x axis for a value of y axis depending on other data plotting

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Your data
chem = {
    "x": [0.96285, 0.98582, 1.01338, 1.09120, 1.12983],
    "concentration": [2, 4, 6, 8, 10]
}

df = pd.DataFrame(chem)

# Interpolate to find concentration at x=1.02
concentration_at_1_02 = np.interp(1.02949, df["x"], df["concentration"])

print(f"Concentration at x=1.02: {concentration_at_1_02}")

# Plot with concentration on x-axis, x on y-axis
plt.plot(df["concentration"], df["x"], linestyle='-', marker='o', markersize=5, label='Original data')

# Mark the interpolated point
plt.plot(concentration_at_1_02, 1.02, marker='*', markersize=12, color='red', label=f'Interpolated: x=1.02')

plt.xlabel("concentration")
plt.ylabel("x")
plt.legend()
plt.grid(True)
plt.show()
```
