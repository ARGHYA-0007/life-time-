```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Your data
chem = {
    "x": [0.96, 0.98, 1.00, 1.025, 1.04],
    "concentration": [2, 4, 6, 8, 10]
}

df = pd.DataFrame(chem)

# Interpolate to find concentration at x=1.02
concentration_at_1_02 = np.interp(1.02, df["x"], df["concentration"])

print(f"Concentration at x=1.02: {concentration_at_1_02}")

# Plot with concentration on x-axis, x on y-axis
plt.plot(df["concentration"], df["x"], linestyle='-', marker='o', markersize=5, label='Original data')

# Mark the interpolated point
plt.plot(concentration_at_1_02, 1.02, marker='*', markersize=12, color='red', label=f'Interpolated: x=1.02')

plt.xlabel("concentration")
plt.ylabel("x")
plt.legend()
plt.grid(True)
plt.show()
```
